# How many darts does it take to find π?

Here's a strange claim: you can find π — the same 3.14159… from every
circle you've ever measured — by throwing darts at random. Not
calculating it. Throwing darts. Actual randomness, turned into one of
the most precise constants in mathematics.

It sounds like a trick. It isn't — it's a real technique called
**Monte Carlo estimation**, used for genuinely hard problems where an
exact formula doesn't exist or is too slow to compute. This is the
simplest possible example of it, small enough to build and understand
completely in one sitting.

## The idea, before the code

Picture a 1×1 square, and a quarter-circle drawn inside it, radius 1,
centred on one corner. The quarter-circle's area is $\frac{\pi}{4}$
(a quarter of $\pi r^2$ with $r=1$); the square's area is exactly $1$.
So the *fraction* of the square covered by the quarter-circle is
$\frac{\pi}{4}$ too.

Now throw darts at the square completely at random. Each one either
lands inside the quarter-circle or it doesn't. If your darts are
genuinely random and you throw enough of them, the *fraction that land
inside* should settle close to that same $\frac{\pi}{4}$ — because
that's just how big the target is relative to the board. Rearrange
that, and:

$$\pi \approx 4 \times \frac{\text{darts inside the quarter-circle}}{\text{darts thrown}}$$

That's the whole idea. Everything below is throwing darts.

## One dart

A "dart" is just a random point — an `x` and a `y`, each somewhere
between 0 and 1. It lands inside the quarter-circle exactly when its
distance from the corner is 1 or less, which (Pythagoras, again) means
$x^2 + y^2 \le 1$.

```python
import random

x, y = random.random(), random.random()
inside = x**2 + y**2 <= 1
print(f"dart at ({x:.3f}, {y:.3f}) — inside the quarter-circle: {inside}")
```

Run that cell a few times — Shift+Enter, repeatedly. Different point
every time, sometimes inside, sometimes not, exactly as you'd expect
from something random. One dart tells you almost nothing about π. A
thousand might. Let's throw a thousand.

```python
def throw_darts(count):
    """Throw `count` random darts at the unit square, and report how
    many landed inside the quarter-circle."""
    inside_count = 0
    for _ in range(count):
        x, y = random.random(), random.random()
        if x**2 + y**2 <= 1:
            inside_count += 1
    return inside_count

darts_thrown = 1000
darts_inside = throw_darts(darts_thrown)
pi_estimate = 4 * darts_inside / darts_thrown
print(f"{darts_inside} of {darts_thrown} darts landed inside")
print(f"estimate of pi: {pi_estimate}")
```

Somewhere in the neighbourhood of 3.14, probably — but run the cell
again and it'll land on a slightly different number each time, since
every run throws a fresh set of random darts. That wobble *is* the
interesting part: an estimate, not a calculation, and estimates carry
uncertainty that shrinks as you gather more evidence.

## Does more darts mean a better estimate?

The claim above was "throw enough of them." Let's actually check that,
instead of taking it on faith — throw increasingly large batches of
darts and watch the estimate as it goes.

```python
import matplotlib.pyplot as plt
import math

dart_counts = [10, 30, 100, 300, 1000, 3000, 10000, 30000]
estimates = [4 * throw_darts(n) / n for n in dart_counts]

plt.plot(dart_counts, estimates, marker="o", label="Monte Carlo estimate")
plt.axhline(math.pi, color="gray", linestyle="--", label="actual pi")
plt.xscale("log")
plt.xlabel("Darts thrown (log scale)")
plt.ylabel("Estimate of pi")
plt.legend()
plt.title("More darts, less wobble")
plt.show()
```

`plt.xscale("log")` matters here — without it, the huge gap between 10
darts and 30,000 would squash every early point into the left edge of
the chart. On a log scale, equal *ratios* (10 to 100, 100 to 1000) get
equal space, which is exactly what you want when a quantity grows by
multiplying rather than adding.

What the chart should show: wild swings with only a handful of darts,
settling down and hugging the dashed line more tightly as the dart
count climbs. That settling-down is the whole method, working — more
random evidence, less uncertainty, same as a poll getting more accurate
with a bigger sample.

## Looking back

Nothing above needed calculus, or a table of digits, or any special
knowledge of $\pi$ beyond "circle stuff." Just geometry (how big is
the target compared to the board), a source of randomness, and the law
of large numbers — the same principle insurance companies, pollsters,
and video game weather systems all lean on, showing up here in its
simplest possible form. Monte Carlo methods scale from this dartboard
all the way up to simulating particle physics and pricing financial
options — the core idea never really changes, only what's being
thrown and what counts as "inside."
